# Concavity in Cubic Consumption Functions

With `CubicBool=True`, HARK builds the consumption function by cubic Hermite interpolation through the endogenous gridpoints, using the exact marginal propensity to consume (MPC) at each one. The model's exact consumption function is concave, so its MPC never rises with wealth. A cubic through the gridpoints need not be concave, though.

This notebook shows where the cubic violates concavity when the expectation over the income shocks is a sum over discrete atoms, and how taking the expectation against the shocks' continuous distributions, `IncShkExpectation="continuous"`, removes the violations. It uses HARK's default parameters throughout.

```python
from unittest import mock

import matplotlib.pyplot as plt
import numpy as np

from HARK.ConsumptionSaving.ConsIndShockModel import (
    IndShockConsumerType,
    get_mNrm_kinks,
)
```

## Measuring concavity violations

A concave consumption function has an MPC that never rises with wealth. The function below checks each segment of a solved cubic consumption function between its gridpoints, from a lower point `m_lo` up: how many segments have an MPC that rises somewhere inside them, and the largest rise.

```python
def cubic_part(solution):
    # The consumption function is the lower envelope of the cubic and the constraint
    return solution.cFunc.functions[0]


def concavity_violations(solution, m_lo, points_per_segment=50):
    cubic = cubic_part(solution)
    knots = np.asarray(cubic.x_list)
    count, largest = 0, 0.0
    for lo, hi in zip(knots[:-1], knots[1:]):
        if hi <= m_lo:
            continue
        m = np.linspace(max(lo, m_lo), hi, points_per_segment)
        mpc = cubic.derivative(m)
        rise = np.max(np.maximum.accumulate(mpc[::-1])[::-1] - mpc)
        if rise > 1e-12:
            count += 1
            largest = max(largest, rise)
    return count, largest


def solve_over_atoms(**parameters):
    # The default: the expectation over the income shocks is a sum over their atoms.
    # HARK refuses CubicBool=True there (check_cubic_pairing), for the reason this
    # notebook shows; to show it, the check is switched off for these solves.
    with mock.patch(
        "HARK.ConsumptionSaving.ConsIndShockModel.check_cubic_pairing",
        lambda *args, **kwargs: None,
    ):
        agent = IndShockConsumerType(cycles=0, CubicBool=True, **parameters)
        agent.solve()
    return agent.solution[0]


def solve_continuous(**parameters):
    agent = IndShockConsumerType(
        cycles=0, CubicBool=True, IncShkExpectation="continuous", **parameters
    )
    agent.solve()
    return agent.solution[0]
```

## Over discrete atoms

The default `IndShockConsumerType` takes the expectation over its income shocks as a sum over discrete atoms: seven equiprobable points for each shock, plus the unemployment event. Each atom sends next period's kink, where its borrowing constraint stops binding, back to one point of this period's consumption function, the kink's image under that atom. There the exact solution has a kink of its own. A cubic segment that holds one of these points cannot follow the kink, and its MPC rises on part of the segment. On a coarse grid the segments are wide and the cubic happens to stay concave; on finer grids it does not, and refining the grid does not remove the violations.

This is why HARK refuses `CubicBool=True` over atoms, with an error that explains it: cubic interpolation needs `IncShkExpectation="continuous"`. To show what that prevents, the solves over atoms in this notebook switch the check off.

Below, the model is solved on three grids, and each table row counts the segments above this period's kink whose MPC rises.

```python
grids = [48, 200, 500]
atoms = {count: solve_over_atoms(aXtraCount=count) for count in grids}

print("gridpoints   segments above the kink   violating   largest MPC rise")
for count, solution in atoms.items():
    kink = get_mNrm_kinks(solution)[0]
    knots = np.asarray(cubic_part(solution).x_list)
    n, rise = concavity_violations(solution, kink)
    print(f"{count:10d}   {np.sum(knots > kink):23d}   {n:9d}   {rise:.1e}")
```

The figure shows the MPC on the finest grid around the stretches where it rises, each a short climb where a segment holds one of the kink's images.

```python
solution = atoms[500]
kink = get_mNrm_kinks(solution)[0]
m = np.linspace(kink, 10.0, 400001)
mpc = cubic_part(solution).derivative(m)
rising = m[1:][np.diff(mpc) > 1e-12]
m = np.linspace(rising.min() - 0.05, rising.max() + 0.05, 20001)
plt.plot(m, cubic_part(solution).derivative(m))
plt.xlabel("market resources m")
plt.ylabel("MPC")
plt.title("The cubic's MPC where it rises, over atoms, 500 gridpoints")
plt.show()
```

## The continuous expectation

Taking the expectation against the shocks' continuous distributions removes the atoms' images: next period's kink then leaves no kink in this period's consumption function, and the cubic has nothing it cannot follow. With `IncShkExpectation="continuous"`, the solver integrates by Gauss-Legendre quadrature in pieces, cut at each unit step of the shocks' standard-normal coordinates and wherever next period's cash-on-hand reaches its kink, so that the integrand is smooth on each piece. The continuous expectation needs income shocks that are bounded below. By default it takes each as a jumpless truncated lognormal (`LogNormalTruncatedJumpless`): a lognormal truncated below at three standard deviations of its log, whose density is zero at that bound and has no upper bound. Over atoms the default is the lognormal truncated at three standard deviations on both sides, whose density jumps at its bounds; against a continuous distribution, such a jump would leave a jump in the second derivative of the consumption function where next period's kink is reached through the bound.

```python
continuous_grids = [48, 200]
continuous = {count: solve_continuous(aXtraCount=count) for count in continuous_grids}

print("gridpoints   segments above the kink   violating   largest MPC rise")
for count, solution in continuous.items():
    kink = get_mNrm_kinks(solution)[0]
    knots = np.asarray(cubic_part(solution).x_list)
    n, rise = concavity_violations(solution, kink)
    print(f"{count:10d}   {np.sum(knots > kink):23d}   {n:9d}   {rise:.1e}")
```

The cubic through these gridpoints is concave on both grids.

## The natural borrowing constraint

With no artificial constraint (`BoroCnstArt=None`), the natural one binds: the most a consumer could borrow and still repay out of the worst possible income, which the bounded distributions make finite. Consumption then falls to zero at that limit with no kink, and the MPC falls from one just above it. Where a segment of the cubic would still not be concave there, the solver first takes that segment's expectations more accurately, then adds gridpoints inside it; each solution records where, in `ConcavityRefinement`.

```python
natural = solve_continuous(BoroCnstArt=None)
limit = natural.mNrmMin
n, rise = concavity_violations(natural, limit)
refinement = natural.ConcavityRefinement
print(f"natural limit: {limit:.4f}")
print(f"segments whose MPC rises: {n}, largest rise {rise:.1e}")
print(
    f"segments taken more accurately: {np.count_nonzero(refinement['level'])}, "
    f"segments with gridpoints added: {np.count_nonzero(refinement['depth'])}"
)

d = np.geomspace(1e-6, 20.0, 2001)
plt.semilogx(d, natural.cFunc.derivative(limit + d))
plt.xlabel("market resources above the natural limit")
plt.ylabel("MPC")
plt.title("The MPC above the natural limit, continuous expectation")
plt.show()
```

## The atoms' own error

Atoms also make an error of their own: they replace the shocks' distributions by a few points, and refining the asset grid does not remove it. The comparison below solves the same model, with the same distributions (the jumpless truncated lognormals that are the continuous expectation's default), over seven atoms a shock and continuously, and measures the largest difference in consumption on each grid.

```python
jumpless = {"PermShk_tails": "jumpless", "TranShk_tails": "jumpless"}
m = np.linspace(0.0, 20.0, 2001)
print("gridpoints   max |c(atoms) - c(continuous)|")
for count in continuous_grids:
    gap = np.max(
        np.abs(
            solve_over_atoms(aXtraCount=count, **jumpless).cFunc(m)
            - continuous[count].cFunc(m)
        )
    )
    print(f"{count:10d}   {gap:.1e}")
```

The gap is about the same on both grids: it is the atoms', not the grid's.

## Nodes a piece

The continuous expectation's accuracy is set by the grid and by the number of Gauss-Legendre nodes in each piece of the integral: `PermShkQuadCount` for the permanent shock and `TranShkQuadCount` for the transitory one, 4 each by default. On the default grid, the table compares solutions at two to four nodes a piece with one at eight.

```python
reference = solve_continuous(PermShkQuadCount=8, TranShkQuadCount=8).cFunc
print("nodes a piece   max |c - c(8 nodes)|")
for nodes in [2, 3, 4]:
    cFunc = solve_continuous(PermShkQuadCount=nodes, TranShkQuadCount=nodes).cFunc
    print(f"{nodes:13d}   {np.max(np.abs(cFunc(m) - reference(m))):.1e}")
```

The error falls geometrically in the nodes, by more than an order of magnitude for each node more.
